```python
# Import libraries
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType, DecimalType, DateType
```

```python
# Define schema and merge keys for silver tables

silver_table_schema = StructType([
    StructField('indicator_id', IntegerType(), False),
    StructField('indicator_name', StringType(), False),
    StructField('magnitude_id', IntegerType(), False),
    StructField('magnitude_name', StringType(), False),
    StructField('datetime_locale', TimestampType(), False),
    StructField('date_locale', DateType(), False),
    StructField('hour_locale', IntegerType(), False),
    StructField('datetime_utc', TimestampType(), False),
    StructField('geo_id', IntegerType(), False),
    StructField('geo_name', StringType(), False),
    StructField('value', DecimalType(10,4), False),
    StructField('ingestion_timestamp', TimestampType(), False)    
])

silver_merge_keys = ['indicator_id','magnitude_id','datetime_utc','geo_id']
```

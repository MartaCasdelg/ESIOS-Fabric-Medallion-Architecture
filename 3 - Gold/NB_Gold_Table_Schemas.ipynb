# Schemas

```python
# Import libraries
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType, DecimalType, DateType, BooleanType
```

## Fact Tables

```python
gold_fact_demand_schema = StructType([
    StructField('datetime_locale', TimestampType(), False),
    StructField('date_locale', DateType(), False),
    StructField('hour_locale', IntegerType(), False),
    StructField('datetime_utc', TimestampType(), False),
    StructField('ingestion_timestamp', TimestampType(), False),
    StructField('forecasted_demand', IntegerType(), True),
    StructField('real_demand', IntegerType(), True),    
    StructField('scheduled_demand', IntegerType(), True),    
    StructField('total_scheduled_demand', IntegerType(), True)       
])

gold_fact_demand_indicator_map = {
    '544': 'forecasted_demand',
    '1293': 'real_demand',
    '545': 'scheduled_demand',
    '1941': 'total_scheduled_demand'
}

gold_fact_demand_merge_keys = ['datetime_utc']
```

```python
gold_fact_market_schema = StructType([
    StructField('datetime_locale', TimestampType(), False),
    StructField('date_locale', DateType(), False),
    StructField('hour_locale', IntegerType(), False),
    StructField('datetime_utc', TimestampType(), False),
    StructField('ingestion_timestamp', TimestampType(), False),
    StructField('geo_id', IntegerType(), False),
    StructField('magnitude_id', IntegerType(), False),
    StructField('market_price', DecimalType(10,4), False)  
])

gold_fact_market_indicator_map = {
    '600': 'market_price'
}

gold_fact_market_merge_keys = ['datetime_utc', 'geo_id', 'magnitude_id']
```

```python
gold_fact_generation_schema = StructType([
    StructField('datetime_locale', TimestampType(), False),
    StructField('date_locale', DateType(), False),
    StructField('hour_locale', IntegerType(), False),
    StructField('datetime_utc', TimestampType(), False),
    StructField('ingestion_timestamp', TimestampType(), False),
    StructField('indicator_id', IntegerType(), False),
    StructField('magnitude_id', IntegerType(), False), 
    StructField('value', IntegerType(), False)
])

gold_fact_generation_merge_keys = ['datetime_utc', 'indicator_id', 'magnitude_id']

# No traspose, so no need for indicator map
```

```python
gold_fact_generation_forecast_schema = StructType([
    StructField('datetime_locale', TimestampType(), False),
    StructField('date_locale', DateType(), False),
    StructField('hour_locale', IntegerType(), False),
    StructField('datetime_utc', TimestampType(), False),
    StructField('ingestion_timestamp', TimestampType(), False),
    StructField('indicator_id', IntegerType(), False),
    StructField('magnitude_id', IntegerType(), False), 
    StructField('value', IntegerType(), False) 
])

gold_fact_generation_forecast_merge_keys = ['datetime_utc', 'indicator_id', 'magnitude_id']

# No traspose, so no need for indicator map
```

## Dim Tables

```python
gold_dim_indicators_schema = StructType([
    StructField('indicator_id', IntegerType(), False),
    StructField('indicator_name', StringType(), False), 
    StructField('esios_indicator_short_name', StringType(), False), # Short Name from metadata
    StructField('indicator_display_name', StringType(), False), # indicator_display_name from metadata
    StructField('resource_type', StringType(), False) # Renewable or non-renewable
])
```

```python
gold_dim_country_schema = StructType([
    StructField('geo_id', IntegerType(), False), # geo_id from silver.market
    StructField('geo_name', StringType(), False), # geo_name from silver.market
    StructField('country_code', StringType(), False), # Country code created from mapping in gold NB
    StructField('flag', StringType(), False) # URL image created from country_code
])
```

# 

```python
gold_dim_magnitude_schema = StructType([
    StructField('magnitude_id', IntegerType(), False), # magnitude_id from silver.generation, silver.generation_forecast and silver.market
    StructField('magnitude_name', StringType(), False) # magnitude_name from silver.generation, silver.generation_forecast and silver.market
])
```

```python
gold_dim_calendar_schema = StructType([
    StructField('date', DateType(), False),
    StructField('day_of_week', IntegerType(), False),
    StructField('day_name', StringType(), False),
    StructField('is_weekend', BooleanType(), False),
    StructField('week_of_year', IntegerType(), False),
    StructField('month_number', IntegerType(), False),
    StructField('month_name', StringType(), False),
    StructField('month_short_name', StringType(), False),
    StructField('quarter_number', IntegerType(), False),
    StructField('quarter_name', StringType(), False),
    StructField('year', IntegerType(), False),
    StructField('year_month', StringType(), False),
    StructField('year_month_number', IntegerType(), False),
])
```

```python
gold_dim_hour_schema = StructType([
    StructField('hour_key', IntegerType(), False),
    StructField('hour_label', StringType(), False),
    StructField('hour_range', StringType(), False),
    StructField('day_period', StringType(), False),
    StructField('day_period_key', IntegerType(), False)
])
```
